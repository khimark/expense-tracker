## Project Title

PERSONAL EXPENSE TRACKER

### Problem Statement


Individuals in Kenya for example students, spend money on food, transport, airtime, and data without trackit it. As a result, by the end of the month, they end up not knowing where the money went thus they end up complaining that their money is running out fast.

This project proposes a personal expense tracker system that allows users record, review, and acknowledge their expenses and areas that need regulating.


### Project Objectives

General Objective: To develop a personal expense tracker for tracking the individual spending habit

Specific Objectives:

- Allows user's to record their expenses by; adding, list,update, and delete their expenses.
- Flags expenses beyond a certain limit
- User accesses a summary showing the total, average, and total per category of their expenses.

### Proposed Solution

A web-based application developed using Python and SQLite. Users will record their transactions, and view a summary of their financial expense.

### Proposed Application Functionalities

1. Add Expense - Add a new expense with title, amount, category, data, note
2. View Expense - Show all exppenses in a table
3. Update Expense -  Edit/Change an existing expense
4. Delete Expense - Remove an expense from the database
5. Summary - Show total spent, average, count, and total per category
